```python
data['Amount'] = data['Amount'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)
country_wise_sales = data.groupby("Country")["Amount"].sum()
country_wise_sales
```

```python
sales_person = data.groupby("Sales Person")["Amount"].sum()
sales_person
```

```python
Product_sales = data.groupby("Product")["Amount"].sum()
Product_sales.sort_values(ascending=False)
Product_sales
```

```python
data['Date'] = pd.to_datetime(data['Date'], format='%d/%m/%Y')
```

```python
data['month'] = data['Date'].dt.month
```

```python
data['year'] = data['Date'].dt.year
```

```python
import matplotlib.pyplot as plt
months = ['jan','feb','march','april','may','june']
energy_consumption = [ 1200, 3400, 2500, 1000, 2600, 1800]
plt.plot(months,energy_consumption , color='b',marker='o' ,linestyle="--")
plt.title("energy consumption over 6 months")
plt.xlabel("months")
plt.ylabel("energy consumption")
plt.show()
```

```python
# Pie chart to show energY consumtion distribution
plt.pie(energy_consumption, labels=months, autopct='%1.1f%%', colors=['gold', 'lightcoral', 'lightskyblue', 'lightgreen', 'pink', 'lavender'])

# Add a title
plt.title('Energy Consumption Share by Source')

# Display the pie chart
plt.show()
```
